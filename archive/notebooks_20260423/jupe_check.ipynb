```python
from os import mkdir
from pathlib import Path
from typing import Any

is_path = "C:/Users/goku/Downloads/"
source_dir = Path(is_path).iterdir()
sorting_cat = {'extensions': ['wav', 'mp3']}
out_path = is_path + 'audio_sorted' + '1'
# mkdir(out_path)

def file_map(path: Path, category_rules: dict[str, list[str]]) -> Any:
    print(f'PATH: {path}')

    # TODO 1: Read the file extension from `path`.
    file_ext = path.suffix
    print(f'first file ext {file_ext}')

    # TODO 2: Convert it to lowercase.
    file_ext = file_ext.lower()

    # TODO 3: Remove the leading dot if needed.
    file_ext = file_ext[1:]
    print(f'file_ext: {file_ext}')

    # TODO 4: Loop through `category_rules` and find the matching category.
    for category, extensions in category_rules.items():
        if file_ext in extensions:
            return category, path

    # TODO 6: If no match is found, return "misc".
    return "misc", path

for s_path in source_dir:
    cat, path = file_map(path=s_path, category_rules=sorting_cat)
    print(f'Category: {cat}')  # Added to see results




```
